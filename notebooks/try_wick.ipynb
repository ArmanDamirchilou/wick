# Try wick

[wick](https://github.com/ArmanDamirchilou/wick) answers questions about a PDF using a language model that runs on your own machine — no API keys, no account, no internet.

It's a command-line tool. This notebook exists so you can see what it does before installing anything: it installs the real package and runs it, here, on a free machine. Run the cells in order; the whole thing takes about three minutes, most of it downloading the model once.

````python
%pip install --quiet "wick-offline[embeddings] @ git+https://github.com/ArmanDamirchilou/wick@main"
%pip install --quiet llama-cpp-python --extra-index-url https://abetlen.github.io/llama-cpp-python/whl/cpu
print("installed")
````

## The one step that needs a connection

`--download-model` fetches a 1.1 GB language model and the retrieval encoder. Everything after this runs locally — on your own laptop that means you can unplug the network and it keeps working.

````python
!wick --download-model
````

## Ask an example document

A short earthquake-safety guide. The last question isn't answered anywhere in it — watch what happens.

````python
!wget -q https://raw.githubusercontent.com/ArmanDamirchilou/wick/main/examples/earthquake-safety.pdf

!wick earthquake-safety.pdf "What should I do if I am trapped under rubble?"
!wick earthquake-safety.pdf "How much water should I store, and for how long?"
!wick earthquake-safety.pdf "What is the population of Tokyo?"
````

That last one is the point. A similarity threshold rejects the question before the model ever sees it, so a small model can't quietly answer from its training data.

## Show your working

`--show-sources` prints the passages the answer was drawn from, so you can check it against the document instead of trusting it.

````python
!wick earthquake-safety.pdf "How much water should I store?" --show-sources
````

## A document that isn't in English

Retrieval is a separate model from the one that writes the answer, and the default only understands English. wick checks the document's script and switches to a multilingual encoder itself — the only thing left to choose is a language model that speaks the language.

Gemma 3n is 3 GB, so this cell takes a few minutes.

````python
!wget -q https://raw.githubusercontent.com/ArmanDamirchilou/wick/main/examples/water-cycle-fa.pdf
!wick --download-model --model-name gemma-3n-e2b

!wick water-cycle-fa.pdf "بیشتر آب کره زمین کجاست؟" --model-name gemma-3n-e2b
````
